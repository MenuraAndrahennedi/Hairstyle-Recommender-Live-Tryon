# Prepare Kept Asset Labeling

Create a dedicated normalized-labeling template for the kept hairstyle asset bank.

```python
from pathlib import Path
import sys

def find_project_root(start: Path | None = None) -> Path:
    current = (start or Path.cwd()).resolve()
    for candidate in [current, *current.parents]:
        if (candidate / 'backend').exists() and (candidate / 'notebooks').exists():
            return candidate
    raise RuntimeError('Could not locate project root.')

PROJECT_ROOT = find_project_root()
BACKEND_ROOT = PROJECT_ROOT / 'backend'
if str(BACKEND_ROOT) not in sys.path:
    sys.path.append(str(BACKEND_ROOT))

PROJECT_ROOT
```

```python
import pandas as pd

from app.ml.celeba_hair_rich import (
    normalized_label_row_from_asset,
    read_jsonl,
)

REVIEWED_ROOT = BACKEND_ROOT / 'data' / 'processed' / 'celeba_hair_rich_assets' / 'reviewed'
KEPT_JSONL = REVIEWED_ROOT / 'kept_assets.jsonl'
LABELING_ROOT = REVIEWED_ROOT / 'labeling'
LABEL_TEMPLATE_CSV = LABELING_ROOT / 'kept_asset_labeling_template.csv'

REVIEWED_ROOT
```

```python
kept_assets = read_jsonl(KEPT_JSONL)
print('Kept assets available for labeling:', len(kept_assets))
kept_assets[0] if kept_assets else None
```

```python
label_rows = [normalized_label_row_from_asset(asset) for asset in kept_assets]
label_frame = pd.DataFrame(label_rows)

LABELING_ROOT.mkdir(parents=True, exist_ok=True)
label_frame.to_csv(LABEL_TEMPLATE_CSV, index=False, encoding='utf-8')

print('Wrote labeling template:', LABEL_TEMPLATE_CSV)
label_frame.head(20)
```
