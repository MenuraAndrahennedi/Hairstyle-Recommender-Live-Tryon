# Prepare Live Segmentation Review

Create a review CSV and editable mask copies for webcam/live segmentation frames. Use this when the pseudo masks are too weak to trust for direct fine-tuning.

```python
from pathlib import Path
import sys
import json
import pandas as pd
from IPython.display import display

def find_project_root(start: Path | None = None) -> Path:
    current = (start or Path.cwd()).resolve()
    for candidate in [current, *current.parents]:
        if (candidate / 'backend').exists() and (candidate / 'notebooks').exists():
            return candidate
    raise RuntimeError('Could not locate project root.')

PROJECT_ROOT = find_project_root()
BACKEND_ROOT = PROJECT_ROOT / 'backend'
if str(BACKEND_ROOT) not in sys.path:
    sys.path.append(str(BACKEND_ROOT))

from app.ml.live_segmentation import LIVE_SEGMENTATION_DATASET_DIR, prepare_live_segmentation_review

PROJECT_ROOT
```

```python
DATASET_DIR = LIVE_SEGMENTATION_DATASET_DIR
print('Dataset dir:', DATASET_DIR)
```

```python
review_payload = prepare_live_segmentation_review(DATASET_DIR)
print('Review CSV:', review_payload['review_csv_path'])
print('Editable mask dir:', review_payload['editable_mask_dir'])
print('Summary path:', review_payload['summary_path'])
print('\nSummary:')
print(review_payload['summary_path'].read_text(encoding='utf-8'))
```

```python
review_df = pd.read_csv(review_payload['review_csv_path'])
print('Rows:', len(review_df))
display(review_df.head(12))
```

## Review instructions

1. Open the editable masks in `editable_mask_path` and fix them where needed.
2. In the CSV, set `review_keep` to `yes` or `no`.
3. Set `mask_fixed` to `yes` if you edited the copied mask.
4. Add optional `review_notes`.
